# 🧠 Introducción a los Algoritmos No Supervisados


Hasta ahora hemos trabajado con algoritmos **supervisados**, donde el modelo aprende a partir de ejemplos **con etiquetas**: sabíamos cuál era la respuesta correcta (como el precio de una casa, o si un cliente se irá o no).

Pero, ¿qué hacemos cuando **no tenemos etiquetas**? ¿Podemos seguir descubriendo patrones en los datos?

✅ **Sí**. Para eso existen los algoritmos **no supervisados**.



## 🎯 ¿Qué son los algoritmos no supervisados?

- Son modelos que **no necesitan etiquetas**.
- Su objetivo es **descubrir estructuras ocultas** en los datos.
- Trabajan **solo con las características (`X`)**, sin una variable objetivo (`y`).



## 🔍 Ejemplos de tareas no supervisadas

| Tarea                     | ¿Qué hace?                                                | Ejemplo en la vida real                   |
|--------------------------|-----------------------------------------------------------|-------------------------------------------|
| **Clustering**           | Agrupa datos similares entre sí                           | Segmentar clientes en marketing           |
| **Reducción de dimensión** | Resume los datos conservando la mayor información posible | Visualización, compresión de imágenes     |
| **Detección de anomalías** | Encuentra datos que se comportan de forma inusual        | Detectar fraudes en transacciones bancarias |



## 📌 Comparación Supervisado vs No Supervisado

|                     | Supervisado                     | No Supervisado                  |

## 🧪 Comenzamos con: **K-Means Clustering**

`K-Means` es un algoritmo muy utilizado para **agrupar datos automáticamente** en grupos que tengan características similares.



# **K-means algoritmo no supervisado**

K-means busca dividir un conjunto de datos en K clusters diferentes de tal manera que los puntos dentro de cada cluster sean más similares entre sí que a los puntos de otros clusters.

Proceso del Algoritmo

- **Inicialización** : Selecciona K puntos (centroides) iniciales aleatoriamente del dataset. Estos puntos serán los centroides iniciales de los clusters.
- **Asignación de Clusters** : Asigna cada punto de los datos al cluster cuyo centroide sea el más cercano.
- **Actualización de Centroides** : Calcula la media de los puntos asignados a cada cluster para actualizar los centroides.
Repetición: Repite los pasos 2 y 3 hasta que los centroides ya no cambien significativamente o se alcance un número máximo de iteraciones.

### **Aplicaciones del K-means**

K-means tiene una variedad de aplicaciones, incluyendo:

- **Segmentación de clientes** : Agrupar clientes en función de sus características y comportamientos.


- **Compresión de imágenes** : Reducir la cantidad de colores en una imagen.

- **Análisis de patrones** : Identificar patrones y estructuras en datos complejos.

### **Paso 3: Seleccionar las características relevantes**

```python
# Renombrar las columnas para simplificar
df.rename(columns={'Annual Income (k$)': 'Annual_Income', 'Spending Score (1-100)': 'Spending_Score'}, inplace=True)

# Seleccionar las características 'Annual_Income' y 'Spending_Score'
X = df[['Annual_Income', 'Spending_Score']]
display(X.head())
```

### **Paso 4: Escalar los datos**


Escalar los datos es una etapa importante en el preprocesamiento de datos, especialmente para algoritmos de machine learning como K-means, que son sensibles a la escala de las características. Escalar los datos asegura que todas las características contribuyan igualmente a la distancia calculada entre los puntos.

### **¿Por qué escalar los datos?**

- **Uniformidad** : Cuando las características tienen diferentes rangos, las características con rangos más grandes pueden dominar el cálculo de distancias. Escalar los datos uniformiza el rango de todas las características.

- **Mejora del rendimiento** : Escalar los datos puede mejorar la convergencia y la precisión de los algoritmos de clustering y clasificación.




Se crea una instancia de StandardScaler. Este objeto se utilizará para estandarizar las características de los datos. La estandarización, también conocida como normalización Z-score, transforma los datos para que tengan una media de 0 y una desviación estándar de 1. Esto es útil para que todas las características tengan la misma escala, lo que puede mejorar el rendimiento de muchos algoritmos de machine learning

```python
# Escalar los datos
scaler = StandardScaler()

```

a continuaciób se realizan dos operaciones en una:

- **Ajustar (fit)** : El método fit calcula la media y la desviación estándar para cada característica en el conjunto de datos X. Esto establece los parámetros necesarios para realizar la estandarización.

- **Transformar (transform)** : El método transform aplica la transformación de estandarización a los datos X utilizando los parámetros (media y desviación estándar) calculados en el paso de ajuste.

```python
# Escalar los datos
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("X ")
print(X)
print("X_scaled")
X_scaled
```

### **Paso 5: Crear y ajustar el modelo K-means**

### **Interpretación del Gráfico del Codo**

Al ejecutar el código siguiente, se generará un gráfico con la inercia en el eje y y el número de clusters en el eje x. El "codo" en el gráfico es el punto donde la tasa de disminución de la inercia se reduce significativamente. Este punto sugiere el número óptimo de clusters.

```python
#Determinar el número óptimo de clusters usando el método del codo
inertia = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
import plotly.graph_objs as go

# Crear gráfico interactivo del método del codo
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=list(range(1, 11)),
    y=inertia,
    mode='lines+markers',
    marker=dict(size=8),
    line=dict(width=2),
    name='Inercia',
    hovertemplate='k = %{x}<br>Inercia = %{y:.2f}<extra></extra>'
))

fig.update_layout(
    title='Método del Codo (Elbow Method)',
    xaxis_title='Número de Clusters (k)',
    yaxis_title='Inercia',
    width=800,
    height=400
)

fig.show()

```

### **Interpretación**

Busca el punto en el que la curva comienza a aplanarse, indicando que agregar más clusters no proporciona una mejora significativa en términos de inercia. Este punto es el "codo" y su valor en el eje x es el número óptimo de clusters.

En el caso del dataset "Mall Customers", normalmente se observa que el "codo" está alrededor de k = 5. Esto sugiere que dividir los clientes en 5 clusters es una elección razonable para este dataset.

### **Paso 6: Ajustar K-means con el número óptimo de clusters**

🔍 ¿Qué es k-means++

Es una técnica de inicialización inteligente de los centroides del algoritmo K-Means.

Fue propuesta para mejorar la velocidad de convergencia y evitar malos resultados por mala elección de los centroides iniciales.

🧠 Problema con K-Means clásico:

- En K-Means "normal", los centroides se eligen al azar. Esto puede causar:

- Convergencia lenta.

- Agrupaciones pobres (malos mínimos locales).

- Resultados inestables entre ejecuciones.

✅ Ventaja de k-means++:

Selecciona los centroides iniciales de manera más informada. El proceso es:

- Selecciona el primer centroide aleatoriamente.

- Para cada punto restante, calcula su distancia al centroide más cercano.

- Elige el siguiente centroide con probabilidad proporcional al cuadrado de la distancia (puntos lejanos tienen más chance).

- Repite hasta tener k centroides.

Esto ayuda a:

- Aumentar la calidad de los clusters.

- Disminuir el número de iteraciones necesarias.

```python
# Crear el modelo K-Means con 5 clusters, usando inicialización k-means++ para mejorar la convergencia,
# y random_state=42 para asegurar resultados reproducibles
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42)


# Ajustar el modelo a los datos escalados
kmeans.fit(X_scaled)

# Obtener las etiquetas de los clusters
labels = kmeans.predict(X_scaled)




# Obtener los centroides
centroids = kmeans.cluster_centers_
print(centroids)
```

### **Paso 7: Visualizar los resultados**

```python
import plotly.graph_objects as go

# Crear el gráfico de dispersión
fig = go.Figure()

# Añadir los puntos de datos con colores por cluster
fig.add_trace(go.Scatter(
    x=X_scaled[:, 0],
    y=X_scaled[:, 1],
    mode='markers',
    marker=dict(
        color=labels,
        colorscale='Viridis',
        size=10,
        opacity=0.8,
        showscale=True,
        colorbar=dict(title='Cluster')
    ),
    name='Clientes',
    text=[f'Cluster: {l}' for l in labels],
    hoverinfo='text+x+y'
))

# Añadir los centroides
fig.add_trace(go.Scatter(
    x=centroids[:, 0],
    y=centroids[:, 1],
    mode='markers',
    marker=dict(
        color='red',
        size=15,
        symbol='x'
    ),
    name='Centroides',
    hoverinfo='x+y+text',
    text=['Centroide']*len(centroids)
))

# Personalizar el diseño
fig.update_layout(
    title='K-means Clustering de Clientes del Centro Comercial (Interactivo)',
    xaxis_title='Ingreso Anual (escalado)',
    yaxis_title='Puntuación de Gasto (escalado)',
    width=800,
    height=600,
    hovermode='closest',
    template='plotly_white'
)

fig.show()
```

```python
import joblib
# Guardar el modelo K-means
joblib.dump(kmeans, 'kmeans_model.pkl')

# Guardar el scaler
joblib.dump(scaler, 'scaler.pkl')
```

```python
import seaborn as sns
# Cargar el modelo K-means y el scaler
kmeans = joblib.load('kmeans_model.pkl')
scaler = joblib.load('scaler.pkl')

# Nuevos datos de clientes
new_data = pd.DataFrame({
    'Annual_Income': [15, 85, 45, 125],
    'Spending_Score': [39, 81, 77, 40]
})

# Generar nuevos datos de clientes (1000 datos aleatorios)
np.random.seed(42)
annual_income = np.random.randint(15, 150, 1000)
spending_score = np.random.randint(1, 100, 1000)
new_data = pd.DataFrame({
    'Annual_Income': annual_income,
    'Spending_Score': spending_score
})

# Escalar los nuevos datos
new_data_scaled = scaler.transform(new_data)

# Predecir los clusters de los nuevos datos
new_clusters = kmeans.predict(new_data_scaled)
new_data['Cluster'] = new_clusters

new_data_scaled
```

```python
import plotly.graph_objects as go

# Crear figura
fig = go.Figure()

# Agregar puntos de clientes por cluster
for cluster_id in sorted(new_data['Cluster'].unique()):
    cluster_data = new_data[new_data['Cluster'] == cluster_id]
    fig.add_trace(go.Scatter(
        x=cluster_data['Annual_Income'],
        y=cluster_data['Spending_Score'],
        mode='markers',
        marker=dict(size=8),
        name=f'Cluster {cluster_id}',
        hovertemplate='Ingreso: %{x}<br>Gasto: %{y}<extra>Cluster</extra>'
    ))

# Agregar centroides
centroids = scaler.inverse_transform(kmeans.cluster_centers_)
fig.add_trace(go.Scatter(
    x=centroids[:, 0],
    y=centroids[:, 1],
    mode='markers',
    marker=dict(size=16, color='black', symbol='x'),
    name='Centroides',
    hovertemplate='Centroide<br>Ingreso: %{x}<br>Gasto: %{y}<extra></extra>'
))

# Configuración del gráfico
fig.update_layout(
    title='Customer Segments (Plotly)',
    xaxis_title='Annual Income (k$)',
    yaxis_title='Spending Score (1-100)',
    width=900,
    height=600
)

fig.show()
```
